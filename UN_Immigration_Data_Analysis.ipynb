# UN Immigration Data Cleaning and Transformation

## Introduction

This project uses immigration data from the United Nations Population Division. The objective is to clean and transform the original Excel dataset into a structured Pandas DataFrame. The process includes identifying the correct header row, selecting relevant columns, handling missing values, converting yearly immigration figures into numeric values, and calculating total immigration from 1980 through 2013.

```python
import pandas as pd

# Load the Excel workbook
file_path = "Canada.xlsx"

# Display the available worksheets
excel_file = pd.ExcelFile(file_path)
print(excel_file.sheet_names)
```

```python
# Read the original UN immigration data
df = pd.read_excel(
    "Canada.xlsx",
    sheet_name="Canada by Citizenship",
    header=None
)

# Display the first 25 rows
display(df.head(25))
```

```python
# Display rows 15 to 25
display(df.iloc[15:26, :12])
```

```python
# Search for the header row
for i, row in df.iterrows():
    values = row.astype(str).tolist()
    if "Type" in values and "Coverage" in values:
        print("Header row:", i)
        print(values[:12])
        break
```

```python
# Find the header row
header_row = None

for i, row in df.iterrows():
    values = row.astype(str).str.strip().tolist()

    if "Type" in values and "Coverage" in values:
        header_row = i
        break

if header_row is None:
    raise ValueError("Could not find the header row.")

# Use that row as the column names
canada = df.iloc[header_row + 1:].copy()
canada.columns = df.iloc[header_row].tolist()

# Remove completely empty rows and columns
canada = canada.dropna(how="all")
canada = canada.dropna(axis=1, how="all")

print(canada.columns.tolist())
display(canada.head())
```

```python
# Rename the original UN columns
canada = canada.rename(columns={
    "OdName": "Country",
    "AreaName": "Continent",
    "RegName": "Region"
})

# List of immigration years
years = list(range(1980, 2014))

# Convert year column names to integers
canada.columns = [
    int(col) if str(col).strip() in
    [str(y) for y in years]
    else col
    for col in canada.columns
]

# Select required columns
required_columns = [
    "Country", "Continent", "Region", "DevName"
] + years

# Check whether all columns exist
missing = [
    col for col in required_columns
    if col not in canada.columns
]

if missing:
    raise ValueError(f"Missing columns: {missing}")

canada_clean = canada[required_columns].copy()

display(canada_clean.head())
```

```python
# Convert yearly immigration data to numeric
for year in years:
    canada_clean[year] = pd.to_numeric(
        canada_clean[year],
        errors="coerce"
    ).fillna(0).astype(int)

# Remove rows without a country
canada_clean = canada_clean.dropna(
    subset=["Country"]
)

# Set Country as the DataFrame index
canada_clean = canada_clean.set_index("Country")

display(canada_clean.head())
```

```python
# Calculate immigration total from 1980 to 2013
canada_clean["Total"] = canada_clean[years].sum(axis=1)

# Display the first five countries
display(canada_clean.head())
```

```python
# Check the structure of the final dataset
print("Number of rows:", canada_clean.shape[0])
print("Number of columns:", canada_clean.shape[1])

print("\nFirst five countries:")
display(canada_clean.head())

print("\nMissing values:")
print(canada_clean.isnull().sum().sum())

# Confirm that Total is calculated correctly
assert (
    canada_clean["Total"]
    == canada_clean[years].sum(axis=1)
).all()

print("\nAll total calculations are correct!")
```


## Conclusion

The original United Nations immigration dataset was transformed into a structured DataFrame containing country names, geographical classifications, annual immigration figures, and total immigration. The resulting dataset is easier to analyze and can support future statistical analysis and data visualization.
